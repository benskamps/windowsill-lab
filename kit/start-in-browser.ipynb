   can find a real signal, and you're ready to pick a star of your own.

To run a cell, click it and press **Shift + Enter**. Run them in order.

```python
# 1. Copy the lab's code here. Nothing is installed on your computer.
# For a star of your own: put your fork's address here and drop --depth 1,
# so the kit can read when your preregistration was first committed.
!git clone --depth 1 -q https://github.com/benskamps/windowsill-lab
%cd windowsill-lab
!pip -q install numpy
```

## 2. Check the setup

Each line should say `ok`. If MAST (NASA's TESS archive) says `NO`, try again
in a few minutes; the archive has outages.

```python
!PYTHONPATH=src python -m lab.planetkit doctor
```

## 3. Write down the question first

Before looking at any data, the kit makes you say what you're searching and
why. This cell writes that down. For a real star you'd commit it to a public
git repository so anyone can see it came before the result. For this
calibration run you don't need to, and the result will remind you it wasn't
committed. That's expected.

```python
!PYTHONPATH=src python -m lab.planetkit prereg 100100827 --sectors 2 \
    --question "Calibration: does my setup re-find WASP-18 b, a known hot Jupiter?" \
    --out prereg-wasp18.md
print(open("prereg-wasp18.md").read())
```

## 4. Run the search

This downloads one TESS sector of WASP-18 and runs the whole ladder: the
search, 256 shuffles of the star's own data to see how often noise looks this
good, the injection test, every vetting gate, the catalog check, and a
placebo. It takes about three minutes.

```python
!PYTHONPATH=src python -m lab.planetkit run --prereg prereg-wasp18.md --download --out receipt-wasp18.json
```

## 5. Look at it

This is the folded light curve: every dip in the sector lined up on top of
each other at the period the search found. A picture, not a verdict.

```python
import glob
from IPython.display import SVG, display
for path in sorted(glob.glob("receipt-wasp18-s*-fold.svg")):
    display(SVG(filename=path))
```

## What you should see

The verdict at the top of step 4 should be **`already-known`**: your search
re-found a signal that is already catalogued. That's a real result, and it
means your setup works.

If it says something else, don't go looking for new stars yet. Check step 2
and run it again.
